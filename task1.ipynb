## Step 1: Import Libraries and Load Dataset

```python
import pandas as pd
```

```python
# Load the Netflix dataset
df = pd.read_csv(r'netflix_titles.csv\netflix_titles.csv')

# Display first few rows
df.head()
```

```python
# Display last few rows
df.tail()
```

## Step 2: Inspect Data

```python
# Overview of columns, data types, and nulls
df.info()
```

```python
# Check missing values
df.isnull().sum()
```

##  Step 3: Handle Missing Values

```python
# Fill missing 'director' and 'cast' with 'Unknown'
df['director'] = df['director'].fillna('Unknown')
df['cast'] = df['cast'].fillna('Unknown')

# Fill 'country' with 'Unknown'
df['country'] = df['country'].fillna('Unknown')

# Fill 'rating' with 'Unknown'
df['rating'] = df['rating'].fillna('Unknown')

# Fill 'duration' with 'Unknown'
df['duration'] = df['duration'].fillna('Unknown')
```

```python
# Fill 'date_added' with mode or leave as NaT(Not a time) 
# by converting it to dattime format and replacing it with mode"""
    
df['date_added'] = pd.to_datetime(df['date_added'] , errors = 'coerce')
```

```python
# Replacing the NaT value with the mode
df['date_added'] = df['date_added'].fillna(df['date_added'].mode()[0])
```

```python
df['date_added'].mode()[0]
```

```python
# Drop any rows with missing titles (important column), here it is not applicable
df = df.dropna(subset=['title'])
```

## Step 4: Remove Duplicates


```python
print(f"Number of duplicate rows: {df.duplicated().sum()}")
df = df.drop_duplicates()
```

## Step 5: Standardize Text Columns



```python
# Lowercase 'type' and clean spaces
df['type'] = df['type'].str.strip().str.lower()

# Title case for country
df['country'] = df['country'].str.title()

print(df['rating'].unique())  # Check unique ratings
df['rating'] = df['rating'].str.upper()  # Standardize to uppercase
df['listed_in'] = df['listed_in'].str.title()  # Standardize genres
```

## Step 6: Convert Date Formats


```python
# Ensure date_added is datetime and converting it backto string (object)
df['date_added'] = pd.to_datetime(df['date_added'], errors='coerce').dt.strftime('%d-%m-%Y')
```

## Step 7: Clean Column Names

```python
# Standardize column names to snake_case
df.columns = df.columns.str.strip().str.lower().str.replace(' ', '-')
print(df.columns)
```

## Step 8: Fix Data Types


```python
# Ensure release_year is integer
df['release_year'] = df['release_year'].astype(int)
```

## Step 9: Check for Outliers (Optional)


```python
print("\nRelease Year Stats:")
print(df['release_year'].describe())
```

## Step 10: Export Cleaned Dataset


```python
df.to_csv('cleaned_netflix_titles.csv' , index=False)
```
